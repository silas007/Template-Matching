```python
import cv2
import tensorflow
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
```

```python
img1 = cv2.imread("/content/boy_image (1).jpg")
```

```python
plt.imshow(img1)
```

```python
boy = cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
plt.imshow(boy)
```

```python
img2 = cv2.imread("/content/boy_crop_template (1).jpg")
```

```python
plt.imshow(img2)
```

```python
face = cv2.cvtColor(img2,cv2.COLOR_BGR2RGB)
plt.imshow(face)
```

```python
boy.shape
```

```python
face.shape
```

```python
template_height,template_width,channels = face.shape
```

```python
template_height
```

```python
template_width
```

```python
channels
```

```python
result = cv2.matchTemplate(boy,face,cv2.TM_CCOEFF_NORMED)
```

```python
plt.imshow(result)
```

```python
min_val,max_val,min_loc,max_loc = cv2.minMaxLoc(result)
```

```python
min_val
```

```python
max_val
```

```python
min_loc
```

```python
max_loc
```

```python
top_left = max_loc
```

```python
bottom_right = (top_left[0]+template_width,top_left[1]+template_height)
bottom_right
```

```python
boy_detected_template = cv2.rectangle(boy,top_left,bottom_right,(225,255,0),25)
```

```python
boy_detected_template
```

```python
plt.imshow(boy_detected_template)
```

```python
result2 = cv2.matchTemplate(boy,face,cv2.TM_SQDIFF)
```

```python
min_val2,max_val2,min_loc2,max_loc2 = cv2.minMaxLoc(result2)
```

```python
plt.imshow(result2)
```

```python
top_left2 = min_loc2
```

```python
bottom_right2 = (top_left2[0]+template_width,top_left2[1]+template_height)
bottom_right2
```

```python
boy_detected_template2 = cv2.rectangle(boy,top_left2,bottom_right2,(0,255,0),50)
```
